```sql
%python
dbutils.widgets.removeAll()
```

```sql
%python
dbutils.widgets.text("storageName", "adlsproject1")
dbutils.widgets.text("catalogo", "catalog_au")
```

```sql
%python

storageName = dbutils.widgets.get("storageName")
catalogo = dbutils.widgets.get("catalogo")
```

### Eliminacion tablas Bronze

```sql
-- DROP TABLES BRONZE
DROP TABLE IF EXISTS ${catalogo}.bronze.order_products__train;
DROP TABLE IF EXISTS ${catalogo}.bronze.products;
```

```sql
%python
## REMOVE DATA (Bronze)
dbutils.fs.rm(f"abfss://bronze@{storageName}.dfs.core.windows.net/order_products_train", True)
dbutils.fs.rm(f"abfss://bronze@{storageName}.dfs.core.windows.net/products", True)
```

### Eliminacion tablas Silver

```sql
-- DROP TABLES SILVER
DROP TABLE IF EXISTS ${catalogo}.silver.products_orders_enriched;
```

```sql
%python
# REMOVE DATA (Silver)
dbutils.fs.rm(f"abfss://silver@{storageName}.dfs.core.windows.net/products_orders_enriched", True)
```

### Eliminacion tablas Golden

```sql
-- DROP TABLES GOLDEN
DROP TABLE IF EXISTS ${catalogo}.golden.kpis_sales_summary_partitioned;
```

```sql
%python
# REMOVE DATA (Golden)
dbutils.fs.rm(f"abfss://golden@{storageName}.dfs.core.windows.net/kpis_sales_summary_partitioned", True)
```
